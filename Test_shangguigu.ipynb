```python
# 计算每日收益率、收益率最高和最低日期、计算波动率（收益的标准差）——销售数据分析
import numpy as np
import pandas as pd

# 生成示例销售数据
prices = pd.Series(
    [102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1, 109.3, 110.2, pd.NA],
    index = pd.date_range('2026-01-01',periods=11, freq='D'))

# 计算每日收益率的函数
def calculate_daily_returns(prices):

    # 每日收益率
    print('=' * 10 + ' shift(1)方法计算每日收益率 ' + '=' * 10)
    prices_per = (prices / prices.shift(1)) -1
    print(f'{prices_per}')
    print('=' * 10 + ' pct_change()方法计算每日收益率 ' + '=' * 10)
    print(prices.pct_change())
    # print(prices.shift(1))

    # 收益率最高和最低日期
    print('=' * 10 + ' 收益率最高日期 ' + '=' * 10)
    prices_sv = prices_per.sort_values(ascending=False)
    # print(prices_sv)
    # print(prices_sv[prices_sv > 0])
    print(prices_sv.idxmax())

    # 如何自定义输出Series中任一值对应的index？
    # # 1. 获取最高收益率对应的纯净日期（第 0 个位置）
    # max_date = prices_sv.index[0]
    # print(max_date)  # 输出: 2026-01-03 00:00:00 (纯净的 Timestamp 标量)

    # # 2. 获取最低收益率对应的纯净日期（倒数第 1 个位置）
    # min_date = prices_sv.index[-1]
    # print(min_date)

    # # 3. 获取任意第 k 名的日期（例如第 3 大的收益率日期）
    # k_date = prices_sv.index[2]
    # print(k_date)

    # # 如果你写了 head(1)，后面加上 [0] 即可提取
    # print(prices_sv.head(1).index[0])

    # 输出纯净文本的index、使用rolling（滑动窗口）函数计算连续增长或下降几个时间段（例，日、周、月）
    idx_prices_sv = prices_per > 0
    # print(idx_prices_sv)

    # 连续两天上涨的日期
    idxlist_prices_sv = prices_per[idx_prices_sv.rolling(2).sum()==2].index.tolist()
    print('=' * 60)
    print(idxlist_prices_sv)


    # 输出纯净文本的日期
    print('=' * 60)
    print(f'连续两天上涨的日期：')
    for idx in idxlist_prices_sv:
        print(f' ·{idx.date()}')

# 调用函数
# calculate_daily_returns(prices)

# 测试df[df.isna()]的用法
print(prices)
print('=' * 10 + ' 测试df[df.isna()]的用法 ' + '=' * 10)
df1 = prices[prices.isna()].index.to_list()
print(df1)
```

```python
# 每小时销售数据分析——获取纯净索引值写法
import numpy as np
import pandas as pd

np.random.seed(42)

# 生成示例销售数据
hour = pd.date_range('2025-01-01 00:00', periods=24, freq='h')
hourly_sales = pd.Series(np.random.randint(0, 100, 24), index=hour)
# per_hour = hourly_sales.resample('D').sum()
# print(per_hour)

# 营业与非营业时间销售额
# print('=' * 10 + ' 营业时间销售额比例 ' + '=' * 10)
# open_sales = (hourly_sales.iloc[8:22].sum() / hourly_sales.sum()).round(2)
# print(f'{open_sales}')
# close_sales = 1-open_sales
# print(f'{close_sales * 100 :.2f}%')

# 计算每小时销售数据的函数
def analyze_hourly_sales(hourly_sales):

    # 每小时销售数据的描述性统计
    print('=' * 10 + ' 每小时销售数据的描述性统计 ' + '=' * 10)
    print(hourly_sales.describe())

    # 销售额最高和最低的时间段
    print('=' * 10 + ' 销售额最高和最低的时间段 ' + '=' * 10)
    max_hour = hourly_sales.idxmax()
    min_hour = hourly_sales.idxmin()
    print(f'销售额最高的时间段: {max_hour}，销售额: {hourly_sales[max_hour]}')
    print(f'销售额最低的时间段: {min_hour}，销售额: {hourly_sales[min_hour]}')

    # 连续两小时销售额超过50的时间段
    print('=' * 10 + ' 连续两小时销售额超过50的时间段 ' + '=' * 10)
    idx_sales = hourly_sales > 50
    idxlist_sales = hourly_sales[idx_sales.rolling(2).sum() == 2].index.tolist()
    for idx in idxlist_sales:
        print(f' ·{idx}')

# 题目要求计算的函数
def per_hour_sales(hourly_sales):
    # 按日计算销售总额
    print('=' * 10 + ' 按日计算销售总额 ' + '=' * 10)
    per_hour = hourly_sales.resample('D').sum()
    print(per_hour)

    # 营业时间与非营业时间的销售额比例
    print('=' * 10 + ' 营业时间与非营业时间的销售额比例 ' + '=' * 10)
    open_sales = (hourly_sales.iloc[8:22].sum() / hourly_sales.sum()).round(2)
    close_sales = 1-open_sales
    print(open_sales,close_sales)

    # 营业额最高的3小时
    print('=' * 10 + ' 营业额最高的3小时 ' + '=' * 10)
    threehours_sales = hourly_sales.rolling(3).sum()
    top3_hours = threehours_sales.sort_values(ascending=False)
    top3_hours_head = top3_hours.head(3)
    print(top3_hours_head)

    # 使用索引输出索引可以得到纯净的索引值
    print('=' * 10 + ' 使用索引输出索引可以得到纯净的索引值 ' + '=' * 10)
    print(f'{top3_hours_head.index[0]}\n{top3_hours_head.index[1]}\n{top3_hours_head.index[2]}')

    # 使用for+items()方法可获取纯净索引值和数据值
    print('=' * 10 + ' 使用for+items()方法可获取纯净索引值和数据值 ' + '=' * 10)
    for idx, value in top3_hours_head.items():
        print(f' ·{idx}，销售额: {value}')

    # .index.to_list()方式输出Timestamp后清洗成纯净索引值
    print('=' * 10 + ' .index以及.index.to_list()方式输出Timestamp后清洗成纯净索引值 ' + '=' * 10)
    # print(type(top3_hours_head.index)) # 输出索引类型为pandas.core.indexes.datetimes.DatetimeIndex
    print(f'{top3_hours_head.index.to_list()}') # 删掉.to_list()后输出的就是Timestamp类型的索引值，效果一样
    for idx in top3_hours_head.index.to_list():
        print(f' ·{idx}')

# 调用函数
# analyze_hourly_sales(hourly_sales)
per_hour_sales(hourly_sales)
```

```python
# 当json各个顶层key对应的值维度不一致时（有列表、字典等），将每个 key 分别转为一个 DataFrame 是最合理且利于后续分析的做法
# 直接创建dataframe时会报错，需先将对应的值转换为列表形式，再创建dataframe
import numpy as np
import pandas as pd
import json

with open('课件及代码/代码/data/test.json','r') as f:
    data = json.load(f)

# pd.set_option('display.max_columns', None) # 显示所有列
# pd.DataFrame(list(data.items()), columns=['Key', 'Value'])

# 各个顶层 key 对应的值类型和维度不一致：
# "users"：列表（包含 3 个用户字典对象）
# "logs"：列表（包含 3 个日志字典对象）
# "settings"：单层字典（配置项对象）
# "metadata"：单层字典（版本及更新时间）

# 动态遍历所有 key，将其转换为对应的 DataFrame 并存入字典，key作为表名
# 与创建一个dataframe的方案一致，pd.DataFrame()中放入字典参数，字典中每个key对应的值用列表存储，
dfs = {}
for k, v in data.items():
    if isinstance(v, list):
        dfs[k] = pd.DataFrame(v) # 创建dataframe用列表作为参数时，列表内每个元素会作为一行数据
    elif isinstance(v, dict):
        dfs[k] = pd.DataFrame([v])  # 单个字典放入列表转为单行数据
    else:
        dfs[k] = pd.DataFrame({k: [v]}) # 创建dataframe用字典作为参数时，每个键值对会作为一列数据出现，键为字段名，值为对应列数据


print(type(dfs['users']))
```

```python
# 房地产数据分析练习
# 导入所需库
import numpy as np
import pandas as pd
from datetime import datetime

# 导入数据
pd_house = pd.read_csv('课件及代码/代码/data/house_sales.csv')

# 数据概览
# pd_house.info()

# 数据清洗
# 删除/填充空值与重复值——需要填充则使用fillna函数
# pd_house['year'] = pd_house.fillna({'year':'unknow'})

# ·1.删除对分析无用的数据——origin_url列
pd_house.drop(columns='origin_url',inplace=True)

# ·2.删除年份为空的数据
# print('=' * 10 + ' 列空值情况(已处理) ' + '=' * 10)
# print(pd_house.isna().sum())
pd_house.dropna(inplace=True)

# ·3.检测与删除重复值
# print('=' * 10 + ' 重复值情况(已处理) ' + '=' * 10)
# print(pd_house.duplicated().sum())
pd_house.drop_duplicates(inplace=True)

# print('\n')
# pd_house.info()

# 数据类型转换

# ·1.字符串列转变为其他处理性能更优列（例，int、float、category、bool，也可采用map函数进行映射）
pd_house['area'] = pd_house['area'].str.replace('㎡','').astype(float)
pd_house['price'] = pd_house['price'].str.replace('万','').astype(float)
pd_house['unit'] = pd_house['unit'].str.replace('元/㎡','').astype(float)
# pd_house['year'] = pd_house['year'].str.replace('年建','').astype(int) 
# 使用.astype('datetime64[s]').dt.to_period('Y')进行转换后，搭配datetime包进行日期计算会报错

# 清理字符串并转为日期类型 (format='%Y' 会默认解析为对应年份的 1月1日)
# errors='coerce' 可以把无法解析的脏数据或缺失值转为 NaT，避免报错
pd_house['year'] = pd.to_datetime(
    pd_house['year'].str.replace('年建', '', regex=False).str.strip(),
    format='%Y',
    errors='coerce'
).dt.year

# ·2.字符串列转变为分类列——检测朝向可否可变成分类
# pd_house['toward'].value_counts()
pd_house['toward'] = pd_house['toward'].astype('category')

# 异常值处理
# 房屋面积异常处理
pd_house = pd_house[(pd_house['area']>20) & (pd_house['area']<600)]

# 房屋售价异常值处理
Q1 = pd_house['price'].quantile(0.25)
Q3 = pd_house['price'].quantile(0.75)

IQR = Q3 - Q1

low_price = Q1 - 1.5*IQR
high_price = Q3 + 1.5*IQR

pd_house = pd_house[(pd_house['price']>low_price) & (pd_house['price']<high_price)]

# 数据特征构造

# ·地区district——地区+街道
# ·1.split拆分——拆分后是由列表组成的Series，需要使用.str作用于每行的元素，并使用[0/1]定位取Series
pd_house['district'] = pd_house['address'].str.split('-').str[0].astype('category')
pd_house['street'] = pd_house['address'].str.split('-').str[1]
# pd_house['district'].value_counts() # 分组汇总得出非常多区，可转变为分类列
pd_house.drop(columns='address',inplace=True)


# ·2.expand参数——此处使用expand参数对'address'列进行了扩列，从1列变成2列，直接赋值给1列会报错，需要使用[0/1]定位取Series
# pd_house['district'] = pd_house['address'].str.split('-', expand=True)[0]
# pd_house['street'] = pd_house['address'].str.split('-', expand=True)[1]

# 楼层分类floor_type
pd_house['floor_type'] = pd_house['floor'].str.split('（').str[0].astype('category')
# pd_house['floor_type'].value_counts() # 分组汇总得出3类楼层——低中高，可转变为分类列
pd_house.drop(columns='floor',inplace=True)

# 是否直辖市city_zxs
# pd_house['city_zxs'] = pd_house['city'].apply( lambda x :1 if x in ['北京','上海','天津','重庆'] else 0)
# 或者用isin()方法
pd_house['city_zxs'] = pd_house['city'].isin(['北京','上海','天津','重庆']).map({ True : 1 , False : 0})
# 随机采样看是否有直辖市的房屋数据
pd_house[pd_house['city_zxs'] == 1].sample(10)

# 卧室数量bedrooms
pd_house['bedrooms'] = pd_house['rooms'].str.split('室').str[0].astype('int')

# 客厅数量livingrooms
# pd_house['livingrooms'] = pd_house['rooms'].str.split('室').str[1].str.replace('厅','').astype(int)
# 字符串的提取处理最好用正则表达式——避免提取的字符出现位置错误
pd_house['livingrooms'] = pd_house['rooms'].str.extract(r'(\d+)厅').astype(int)

# 楼龄b_age
current_year = pd.Timestamp.now().year
pd_house['b_age'] = current_year - pd_house['year']

# 价格分段

# 每平方单价

# pd_house.drop(columns='address',inplace=True)

pd_house.head(5)
# pd_house.info()


# 数据分析
```
