```python
# 学生成绩管理系统
import json
import os
from datetime import datetime

class Student:
    """学生类"""
    
    def __init__(self, student_id, name, age, major):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.major = major
        self.grades = {}  # 存储各科成绩

    def add_grade(self, subject, score):
        """添加成绩"""
        if 0 <= score <= 100:
            self.grades[subject] = score
            return True
        return False
    
    def get_average_grade(self):
        """计算平均成绩"""
        if not self.grades:
            return 0
        return sum(self.grades.values()) / len(self.grades)
    
    def to_dict(self):
        """转换为字典"""
        return {
            'student_id': self.student_id,
            'name': self.name,
            'age': self.age,
            'major': self.major,
            'grades': self.grades
        }

class GradeManager:
    """编程导航学生成绩管理系统"""
    
    def __init__(self, data_file="students.json"):
        self.students = {}
        self.data_file = data_file
        self.load_data()
    
    def add_student(self, student_id, name, age, major):
        """添加学生"""
        if student_id in self.students:
            print(f"学生 {student_id} 已存在")
            return False
        
        student = Student(student_id, name, age, major)
        self.students[student_id] = student  #student实例对象的内存地址，示例：<__main__.Student object at 0x111616a50>
        print(f"成功添加学生: {name}")
        # print(student)
        return True
    
    def add_grade(self, student_id, subject, score):
        """添加成绩"""
        if student_id not in self.students:
            print(f"学生 {student_id} 不存在")
            return False
        
        if self.students[student_id].add_grade(subject, score):
            print(f"成功为学生 {student_id} 添加 {subject} 成绩: {score}")
            return True
        else:
            print("成绩必须在 0-100 之间")
            return False
    
    def get_student_info(self, student_id):
        """获取学生信息"""
        if student_id in self.students:
            student = self.students[student_id]
            print(f"\n学生信息:")
            print(f"学号: {student.student_id}")
            print(f"姓名: {student.name}")
            print(f"年龄: {student.age}")
            print(f"专业: {student.major}")
            print(f"平均成绩: {student.get_average_grade():.2f}")
            print(f"各科成绩: {student.grades}")
            return student
        else:
            print(f"学生 {student_id} 不存在")
            return None
    
    def get_class_statistics(self):
        """获取班级统计信息"""
        if not self.students:
            print("暂无学生数据")
            return
        
        total_students = len(self.students)
        all_averages = [student.get_average_grade() for student in self.students.values()]
        class_average = sum(all_averages) / total_students
        
        print(f"\n班级统计:")
        print(f"总学生数: {total_students}")
        print(f"班级平均分: {class_average:.2f}")
        print(f"最高平均分: {max(all_averages):.2f}")
        print(f"最低平均分: {min(all_averages):.2f}")
    
    def save_data(self):
        """保存数据到文件"""
        data = {sid: student.to_dict() for sid, student in self.students.items()}
        with open(self.data_file, 'w', encoding='utf-8') as f:
            json.dump(data, f, ensure_ascii=False, indent=2)
        print("数据已保存")
    
    def load_data(self):
        """从文件加载数据"""
        if os.path.exists(self.data_file):
            try:
                with open(self.data_file, 'r', encoding='utf-8') as f:
                    data = json.load(f)
                
                for sid, student_data in data.items():
                    student = Student(
                        student_data['student_id'],
                        student_data['name'],
                        student_data['age'],
                        student_data['major']
                    )
                    student.grades = student_data['grades']
                    self.students[sid] = student
                
                print(f"成功加载 {len(self.students)} 个学生数据")
            except Exception as e:
                print(f"加载数据失败: {e}")

# 使用成绩管理系统
def demo_grade_manager():
    manager = GradeManager()
    
    # 添加学生
    manager.add_student("2024001", "程序员鱼皮", 20, "计算机科学")
    manager.add_student("2024002", "编程导航", 19, "软件工程")
    
    # 添加成绩
    manager.add_grade("2024001", "Python", 95)
    manager.add_grade("2024001", "算法", 88)
    manager.add_grade("2024002", "Python", 92)
    manager.add_grade("2024002", "数据结构", 85)
    
    # 查看学生信息
    manager.get_student_info("2024001")
    
    # 查看班级统计
    manager.get_class_statistics()
    
    # 保存数据
    manager.save_data()

demo_grade_manager()
```

```python
# 老鱼简历数据分析器源码——额外增添注释
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

class ResumeDataAnalyzer:
    """老鱼简历数据分析器"""
    
    def __init__(self):
        self.user_data = None
        self.resume_data = None
        self.template_data = None
        self.usage_data = None
        
    def generate_sample_data(self):
        """生成示例数据"""
        np.random.seed(42)
        
        # 生成用户数据
        n_users = 100
        self.user_data = pd.DataFrame({
            'user_id': [f'U{i:04d}' for i in range(1, n_users + 1)],
            'registration_date': pd.date_range(
                start='2023-01-01', 
                end='2024-01-01', 
                periods=n_users
            ),
            'user_type': np.random.choice(['免费用户', 'VIP用户'], n_users, p=[0.8, 0.2]),
            'age': np.random.normal(28, 5, n_users).astype(int),
            'education': np.random.choice(['专科', '本科', '硕士', '博士'], n_users, p=[0.2, 0.5, 0.25, 0.05]),
            'industry': np.random.choice(['IT', '金融', '教育', '制造', '服务'], n_users, p=[0.4, 0.2, 0.15, 0.15, 0.1])
        })
        
        # 生成简历数据
        n_resumes = 2500
        self.resume_data = pd.DataFrame({
            'resume_id': [f'R{i:04d}' for i in range(1, n_resumes + 1)],
            'user_id': np.random.choice(self.user_data['user_id'], n_resumes),
            'template_id': np.random.choice([f'T{i:02d}' for i in range(1, 21)], n_resumes),
            'created_date': pd.date_range(
                start='2023-01-01',
                end='2024-01-01',
                periods=n_resumes
            ),
            'completion_rate': np.random.beta(2, 1, n_resumes) * 100,
            'download_count': np.random.poisson(3, n_resumes),
            'view_count': np.random.poisson(15, n_resumes)
        })
        
        # 生成模板数据
        template_names = [
            '经典简约', '现代商务', '创意设计', '技术专业', '学术研究',
            '销售精英', '金融分析', '教育培训', '医疗健康', '法律专业',
            '市场营销', '人力资源', '项目管理', '数据分析', '产品设计',
            '运营管理', '客服支持', '媒体传播', '艺术创作', '咨询顾问'
        ]
        
        self.template_data = pd.DataFrame({
            'template_id': [f'T{i:02d}' for i in range(1, 21)],
            'template_name': template_names,
            'category': np.random.choice(['通用', '专业', '创意'], 20, p=[0.5, 0.3, 0.2]),
            'price': np.random.choice([0, 9.9, 19.9, 29.9], 20, p=[0.4, 0.3, 0.2, 0.1]),
            'usage_count': np.random.poisson(50, 20)
        })
        
        print("示例数据生成完成")
    
    def user_behavior_analysis(self):
        """用户行为分析"""
        print("\n=== 用户行为分析 ===")
        
        # 用户数据有6列，分别为：user_id, registration_date, user_type, age, education, industry

        # 用户注册趋势
        monthly_registrations = self.user_data.groupby(
            self.user_data['registration_date'].dt.to_period('M')
        ).size()
        
        print(f"月度注册用户趋势:")
        for period, count in monthly_registrations.tail(6).items():
            print(f"  {period}: {count} 位用户")
        
        # 用户类型分布
        user_type_dist = self.user_data['user_type'].value_counts()
        print(f"\n用户类型分布:")
        for user_type, count in user_type_dist.items():
            percentage = count / len(self.user_data) * 100
            print(f"  {user_type}: {count} 人 ({percentage:.1f}%)")
        
        # 用户年龄分布
        age_stats = self.user_data['age'].describe()
        print(f"\n用户年龄统计:")
        print(f"  平均年龄: {age_stats['mean']:.1f} 岁")
        print(f"  年龄范围: {age_stats['min']:.0f} - {age_stats['max']:.0f} 岁")
        
        # 教育背景分布
        education_dist = self.user_data['education'].value_counts()
        print(f"\n教育背景分布:")
        for education, count in education_dist.items():
            percentage = count / len(self.user_data) * 100
            print(f"  {education}: {count} 人 ({percentage:.1f}%)")
    
    def resume_creation_analysis(self):
        """简历创建分析"""
        print("\n=== 简历创建分析 ===")
        
        # 合并用户和简历数据——组合成用户简历数据宽表
        # 用户数据有6列，分别为：user_id, registration_date, user_type, age, education, industry
        # 简历数据有7列，分别为：resume_id, user_id, template_id, created_date, completion_rate, download_count, view_count
        merged_data = self.resume_data.merge(self.user_data, on='user_id')
        
        # 按用户类型分析简历创建数量
        resumes_per_user_type = merged_data.groupby('user_type').agg({
            'resume_id': 'count',
            'completion_rate': 'mean',
            'download_count': 'mean'
        }).round(2)
        
        print(f"按用户类型分析:")
        print(resumes_per_user_type)
        
        # 简历完成度分析
        completion_ranges = pd.cut(
            self.resume_data['completion_rate'],
            bins=[0, 25, 50, 75, 100],
            labels=['0-25%', '26-50%', '51-75%', '76-100%']
        )
        completion_dist = completion_ranges.value_counts()
        
        print(f"\n简历完成度分布:")
        for range_label, count in completion_dist.items():
            percentage = count / len(self.resume_data) * 100
            print(f"  {range_label}: {count} 份 ({percentage:.1f}%)")
        
        # 模板使用分析
        template_usage = self.resume_data['template_id'].value_counts().head(5)
        template_names = self.template_data.set_index('template_id')['template_name']
        
        print(f"\n最受欢迎的模板:")
        for template_id, count in template_usage.items():
            template_name = template_names[template_id]
            print(f"  {template_name} ({template_id}): {count} 次使用")
    
    def template_performance_analysis(self):
        """模板性能分析"""
        print("\n=== 模板性能分析 ===")
        
        # 合并模板和简历数据
        template_performance = self.resume_data.groupby('template_id').agg({
            'resume_id': 'count',
            'completion_rate': 'mean',
            'download_count': 'sum',
            'view_count': 'sum'
        }).round(2)
        
        # 添加模板信息
        template_performance = template_performance.merge(
            self.template_data.set_index('template_id')[['template_name', 'category', 'price']],
            left_index=True,
            right_index=True
        )
        
        # 按使用次数排序
        template_performance = template_performance.sort_values('resume_id', ascending=False)
        
        print(f"模板性能TOP10:")
        for idx, row in template_performance.head(10).iterrows():
            print(f"  {row['template_name']}: 使用{row['resume_id']}次, "
                  f"平均完成度{row['completion_rate']:.1f}%, "
                  f"总下载{row['download_count']}次")
        
        # 按类别分析
        category_performance = template_performance.groupby('category').agg({
            'resume_id': 'sum',
            'completion_rate': 'mean',
            'download_count': 'sum'
        }).round(2)
        
        print(f"\n按类别分析:")
        print(category_performance)
    
    def generate_insights(self):
        """生成数据洞察"""
        print("\n=== 数据洞察与建议 ===")
        
        # 计算关键指标
        total_users = len(self.user_data)
        vip_ratio = len(self.user_data[self.user_data['user_type'] == 'VIP用户']) / total_users
        avg_resumes_per_user = len(self.resume_data) / total_users
        avg_completion_rate = self.resume_data['completion_rate'].mean()
        
        insights = [
            f"用户规模: 平台共有 {total_users} 位用户，VIP用户占比 {vip_ratio:.1%}",
            f"活跃度: 平均每用户创建 {avg_resumes_per_user:.1f} 份简历",
            f"完成度: 简历平均完成度为 {avg_completion_rate:.1f}%",
        ]
        
        # 分析高价值用户特征
        high_activity_users = self.resume_data.groupby('user_id').size()
        top_users = high_activity_users.nlargest(10)
        
        insights.append(f"高活跃用户: 最活跃用户创建了 {top_users.iloc[0]} 份简历")
        
        # 模板优化建议
        low_completion_templates = self.resume_data.groupby('template_id')['completion_rate'].mean()
        poor_templates = low_completion_templates[low_completion_templates < 60]
        
        if len(poor_templates) > 0:
            insights.append(f"优化建议: {len(poor_templates)} 个模板完成度较低，需要优化")
        
        for insight in insights:
            print(f"• {insight}")
    
    def create_visualizations(self):
        """创建可视化图表"""
        print("\n=== 生成数据可视化图表 ===")
        
        # 创建图表
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))
        fig.suptitle('老鱼简历数据分析报告', fontsize=16, fontweight='bold')
        
        # 1. 用户注册趋势
        monthly_data = self.user_data.groupby(
            self.user_data['registration_date'].dt.to_period('M')
        ).size()
        axes[0, 0].plot(range(len(monthly_data)), monthly_data.values, marker='o')
        axes[0, 0].set_title('月度用户注册趋势')
        axes[0, 0].set_xlabel('月份')
        axes[0, 0].set_ylabel('注册用户数')
        axes[0, 0].grid(True, alpha=0.3)
        
        # 2. 用户类型分布
        user_type_counts = self.user_data['user_type'].value_counts()
        axes[0, 1].pie(user_type_counts.values, labels=user_type_counts.index, autopct='%1.1f%%')
        axes[0, 1].set_title('用户类型分布')
        
        # 3. 简历完成度分布
        axes[1, 0].hist(self.resume_data['completion_rate'], bins=20, alpha=0.7, color='skyblue')
        axes[1, 0].set_title('简历完成度分布')
        axes[1, 0].set_xlabel('完成度 (%)')
        axes[1, 0].set_ylabel('简历数量')
        axes[1, 0].grid(True, alpha=0.3)
        
        # 4. 模板使用热力图
        template_usage = self.resume_data.groupby(['template_id', 
                                                  self.resume_data['created_date'].dt.month]).size().unstack(fill_value=0)
        
        if len(template_usage.columns) > 1:
            sns.heatmap(template_usage.iloc[:10], annot=True, fmt='d', cmap='YlOrRd', ax=axes[1, 1])
            axes[1, 1].set_title('模板月度使用热力图')
            axes[1, 1].set_xlabel('月份')
            axes[1, 1].set_ylabel('模板ID')
        
        plt.tight_layout()
        plt.savefig('resume_analysis_report.png', dpi=300, bbox_inches='tight')
        print("可视化图表已保存为 resume_analysis_report.png")
        
        # 显示图表
        plt.show()
    
    def run_complete_analysis(self):
        """运行完整分析"""
        print("=== 老鱼简历数据分析报告 ===")
        print(f"分析时间: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        
        self.generate_sample_data()  # 生成示例数据
        self.user_behavior_analysis() # 用户行为分析
        self.resume_creation_analysis()  # 简历创建分析
        self.template_performance_analysis()  # 模板性能分析
        self.generate_insights()  # 生成数据洞察
        self.create_visualizations()  # 创建可视化图表

# 运行数据分析
analyzer = ResumeDataAnalyzer()
analyzer.run_complete_analysis()
```

```python
# 老鱼简历数据分析器——修改版本
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# 设置字体
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 生成示例数据——包含用户数据、简历数据和模板数据，基本用np库随机生成数据
def generate_sample_data():
    # 固定随机种子
    np.random.seed(42)

    # 生成用户数据——np.random模块搭配生成器生成数据
    # dataframe常用字典形式存储，键为列名，值为数据列表
    n_users = 100
    user_data = pd.DataFrame({
        'user_id': [f'U{i:04d}' for i in range(1, n_users + 1)], # 用户id用列表推导式生成
        'registration_date': pd.date_range(
            start='2023-01-01', 
            end='2024-01-01', 
            periods=n_users
        ),
        'user_type': np.random.choice(['免费用户', 'VIP用户'], n_users, p=[0.8, 0.2]), # 随机选取用户类型匹配上用户id，免费占比0.8，VIP占比0.2
        'age': np.random.normal(28, 5, n_users).astype(int), # 正态分布生成年龄，均值28，标准差5，转换为整数
        'education': np.random.choice(['专科', '本科', '硕士', '博士'], n_users, p=[0.3, 0.4, 0.2, 0.1]),
        'industry': np.random.choice(['IT', '金融', '教育', '制造', '服务'], n_users, p=[0.4, 0.2, 0.1, 0.2, 0.1])
    })
    # print(user_data)
    # print(user_data['registration_date'].dt.to_period('M'))

    # 生成简历数据——np.random模块搭配生成器生成数据
    n_resumes = 250
    resume_data = pd.DataFrame({
        'resume_id': [f'R{i:04d}' for i in range(1, n_resumes + 1)],
        'user_id': np.random.choice(user_data['user_id'], n_resumes),
        'template_id': np.random.choice([f'T{i:02d}' for i in range(1, 21)], n_resumes),
        'created_date': pd.date_range(
            start='2023-01-01',
            end='2024-01-01',
            periods=n_resumes
        ),
        'completion_rate': np.random.beta(2, 1, n_resumes) * 100, #beta分布生成简历完成度，此分布适合生成0-1之间的数值；目前用法偏向1
        'download_count': np.random.poisson(3, n_resumes), # 泊松分布生成下载次数，适合生成非负整数，3为平均值
        'view_count': np.random.poisson(15, n_resumes)
    })
    # print(resume_data)

    # 生成模板数据
    template_names = [
        '经典简约', '现代商务', '创意设计', '技术专业', '学术研究',
        '销售精英', '金融分析', '教育培训', '医疗健康', '法律专业',
        '市场营销', '人力资源', '项目管理', '数据分析', '产品设计',
        '运营管理', '客服支持', '媒体传播', '艺术创作', '咨询顾问'
    ]

    template_data = pd.DataFrame({
        'template_id': [f'T{i:02d}' for i in range(1, 21)],
        'template_name': template_names,
        'category': np.random.choice(['通用', '专业', '创意'], 20, p=[0.5, 0.3, 0.2]),
        'price': np.random.choice([0, 9.9, 19.9, 29.9], 20, p=[0.4, 0.3, 0.2, 0.1]),
        'usage_count': np.random.poisson(50, 20)
    })

    # 用变量接住示例数据并以字典形式返回，方便取用
    return {
        'user_data': user_data,
        'resume_data': resume_data,
        'template_data': template_data
    }

    print("用户示例数据生成完成")

# 用户行为分析——包含用户注册趋势、用户类型分布、用户年龄分布、教育背景分布等分析
def user_behavior_analysis(user_data):
    """用户行为分析"""
    print("\n=== 用户行为分析 ===")

    # 月度注册用户趋势分析——按照注册日期分组聚合并设置为年月并作为索引，然后用size()方法产出的series统计每个年月的用户注册数量，针对单列的聚合统计用size()方法，针对多列的聚合统计用agg()方法
    monthly_registrations = user_data.groupby(user_data['registration_date'].dt.to_period('M')).size()
    # print(monthly_registrations)
    print(type(monthly_registrations))

    # 用for循环遍历series的items()方法，items()方法返回一个可迭代对象，包含索引和值的元组
    print(f"\n月度注册用户趋势:")
    for date_mr,i in monthly_registrations.tail(6).items():
        print(f"{date_mr}: {i} 位用户")

    # 用户类型分布分析
    # 非数值数据分布分析用 value_counts() 方法
    user_type_dist = user_data['user_type'].value_counts() # 统计用户类型数量分布，value_counts()返回series，包含索引和值，索引为用户类型，值为数量
    # print(user_type_dist)
    # print(type(user_type_dist))
    print(f"\n用户类型分布:")
    for user_type, count in user_type_dist.items(): # 惯用的series遍历方法，items()返回一个可迭代对象，包含索引和值的元组
        percentage = count / len(user_data) * 100
        print(f"  {user_type}: {count} 人 {percentage:.1f}%")

    # 用户年龄分布分析
    # 数值数据分布分析用 describe() 方法
    age_stats = user_data['age'].describe() # describe方法统计年龄的描述性统计信息，返回series，包含count、mean、std、min、25%、50%、75%、max等指标
    print(age_stats)
    print(f"\n用户年龄统计:")
    print(f"  平均年龄: {age_stats['mean']:.1f} 岁")
    print(f"  年龄范围: {age_stats['min']:.0f} - {age_stats['max']:.0f} 岁")

    # 教育背景分布
    # 非数值数据分布分析用 value_counts() 方法
    education_dist = user_data['education'].value_counts()
    print(f"\n教育背景分布:")
    for education, count in education_dist.items():
        percentage = count / len(user_data) * 100
        print(f"  {education}: {count} 人 ({percentage:.1f}%)")

# 简历创建分析——包含按用户类型分析简历创建数量、简历完成度分析、模板使用分析等
def resume_creation_analysis(resume_data, user_data, template_data):
    merged_data = resume_data.merge(user_data, on='user_id') # merge方法不会将连接列设置为索引，而是返回一个新的DataFrame，包含所有列

    # 按简历用户分析简历创建数量——agg多列聚合函数，字典内的键为列名，值为聚合类型，groupby函数要配合agg函数使用，因为groupby进行分组，而agg是对分组后多列数据进行聚合
    resumes_per_user_type = merged_data.groupby('user_type').agg({
        'resume_id': 'count',
        'completion_rate': 'mean',
        'download_count': 'mean'
        }).round(2)

    print(f"\n按用户类型分析:resumes_per_user_type")
    print(resumes_per_user_type)

    # 简历完成度分析——cut方法将连续数值数据分段，value_counts()方法统计每个分段的数量，返回的是一个序列
    completion_ranges = pd.cut(
        resume_data['completion_rate'], # 需要分段的连续数值数据
        bins=[0, 25, 50, 75, 100], # bins参数为分段的边界值，左闭右开区间
        labels=['0-25%', '26-50%', '51-75%', '76-100%'] # labels参数为每个分段设置标签，也就是分类
    )
    completion_dist = completion_ranges.value_counts() # values_counts方法统计labels每个分段的数量，返回的是一个序列

    # print(f"\n{type(completion_ranges)}")
    # print(f"\n{completion_dist}")
    print(f"\n简历完成度分布:completion_ranges")
    for range_label, count in completion_dist.items():
        percentage = count / len(resume_data) * 100
        print(f"  {range_label}: {count} 份 ({percentage:.1f}%)")

    # 模板使用分析——value_counts()已自动做倒序排序，能否用merge方法将模板名称和模板id对应起来，便于输出模板名称
    # template_usage = resume_data['template_id'].value_counts().head(5)
    # template_names = template_data.set_index('template_id')['template_name'] # 设置template_id为索引，只取template_name列，返回一个series

    # merge方法合并简历及模板数据后，只展示templatete_id和template_name数据列，value_counts()方法统计每个元组的数量，返回一个series，索引为元组，值为数量
    template_usage = resume_data.merge(template_data, on='template_id')[['template_id','template_name']].value_counts().head(5)

    # print(f"\n{template_usage.index}")
    print(f"\n最受欢迎的模板:template_usage")
    for template_id, count in template_usage.items():
        # template_name = template_names[template_id]
        # print(f"  {template_name} ({template_id}): {count} 次使用")
        
        # 元组索引包含template_id和template_name，所以用template_id[0]和template_id[1]分别取出模板id和模板名称
        print(f" {template_id[0]}-{template_id[1]}: {count} 次使用")

# 模板性能分析——包含按模板分析使用次数、平均完成度、总下载量、总浏览量等指标，并按类别分析模板性能
def template_performance_analysis(resume_data, template_data):
    """模板性能分析"""
    print("\n=== 模板性能分析 ===")
    
    # 合并模板和简历数据——groupby函数要配合agg函数使用，因为groupby进行分组，而agg是对分组后多列数据进行聚合
    template_performance = resume_data.groupby('template_id').agg({
        'resume_id': 'count',
        'completion_rate': 'mean',
        'download_count': 'sum',
        'view_count': 'sum'
    }).round(2)
    
    # 添加模板信息——merge方法将template_performance和template_data按template_id列进行合并，left_index=True表示以左表的索引为基准，right_index=True表示以右表的索引为基准，返回一个新的DataFrame，包含所有列
    # 这里的merge没用用on参数，因为template_performance的template_id是索引，不在列中，而template_data的template_id是列，所以用要么将template_performance的索引重置为列，要么将template_data的template_id设置为索引，这里选择了后者
    template_performance_mg = template_performance.merge(
        template_data.set_index('template_id')[['template_name', 'category', 'price']],
        left_index=True,
        right_index=True
    )
    
    # 按使用次数排序——sort_values方法按指定列排序，ascending=False表示降序排列
    template_performance_sv = template_performance_mg.sort_values('resume_id', ascending=False)

    # print(f'\ngroupby分组聚合后的数据：\n{template_performance}')
    # print(f'\nmerge模板数据后的数据：\n{template_performance_mg}')
    # print(f'\nsort_values排序后的数据：\n{template_performance_sv}')
    
    # iterrows()方法返回一个生成器，生成器每次迭代返回一个元组，包含索引和行数据，行数据是一个Series对象，行数据由字典组成的，字典的键是列名，值是对应的值
    # iterrows()方法适合小数据集，若数据集较大，建议使用itertuples()方法，返回一个命名元组，访问速度更快；可用于dataframe的行遍历，返回一个生成器，每次迭代返回一个命名元组，命名元组的属性名为列名，属性值为对应的值
    # 不能使用items()方法，因为items()方法返回的是一个可迭代对象，包含索引和值的元组，而不是行数据
    print(template_performance_sv.head(10).iterrows())
    print(f"模板性能TOP10:")
    for idx, row in template_performance_sv.head(10).iterrows():
        print(f" {idx}:{row['template_name']}: 使用{row['resume_id']}次, "
                f"平均完成度{row['completion_rate']:.1f}%, "
                f"总下载{row['download_count']}次")
    
    # 按类别分析——groupby函数要配合agg函数使用，因为groupby进行分组，而agg是对分组后多列数据进行聚合
    # groupby函数中的参数会被作为索引，agg函数中的参数会被作为列名，返回一个新的DataFrame，包含字典中指定的列名和聚合函数，round(2)方法将结果保留两位小数
    category_performance = template_performance_sv.groupby('category').agg({
        'resume_id': 'sum',
        'completion_rate': 'mean',
        'download_count': 'sum'
    }).round(2)
    
    print(f"\n按类别分析:")
    print(category_performance)

# 数据洞察与建议——包含用户规模、活跃度、完成度、高活跃用户、模板优化建议等
def generate_insights(user_data, resume_data, template_data):
    """生成数据洞察"""
    print("\n=== 数据洞察与建议 ===")
    
    # 计算关键指标
    total_users = len(user_data) # 总用户数
    vip_ratio = len(user_data[user_data['user_type'] == 'VIP用户']) / total_users # VIP用户占比
    avg_resumes_per_user = len(resume_data) / total_users # 平均每用户创建简历数
    avg_completion_rate = resume_data['completion_rate'].mean() # 简历平均完成度

    # 数据洞察列表
    insights = [
        f"用户规模: 平台共有 {total_users} 位用户，VIP用户占比 {vip_ratio:.1%}",
        f"活跃度: 平均每用户创建 {avg_resumes_per_user:.1f} 份简历",
        f"完成度: 简历平均完成度为 {avg_completion_rate:.1f}%",
    ]
    
    # 分析高价值用户特征
    # 统计每个用户创建的简历数量，size()方法返回一个series，索引为user_id，值为这个id创建的简历数量
    # size()方法与count()方法的区别在于，size()方法统计的是每个分组的大小，包括缺失值，而count()方法统计的是每个分组的非缺失值数量
    # high_activity_users = resume_data.groupby('user_id').size()
    
    # 另一种统计每个用户创建的简历数量的方法，返回一个series，索引为user_id，值为这个id创建的简历数量
    high_activity_users = resume_data['user_id'].value_counts()
    top_users = high_activity_users.head(10) # 取创建简历数量最多的前10个用户

    # iloc方法按位置索引取值，iloc[0]取第一个用户创建的简历数量，loc方法按标签索引取值，loc[top_users.index[0]]取第一个索引值用户创建的简历数量，若为Dataframe则返回一个Series，若为Series则返回一个标量
    insights.append(f"高活跃用户: 最活跃用户创建了 {top_users.iloc[0]} 份简历")
    
    # 模板优化建议
    low_completion_templates = resume_data.groupby('template_id')['completion_rate'].mean()
    # 平均完成度低于60%的模板，按完成度升序排序，如果是Series或者是单列的dataframe则sort_values()不需要额外带参数，不然会报错
    poor_templates = low_completion_templates[low_completion_templates < 60].sort_values()

    # 输出完成度较低的模板信息
    # items()方法返回一个可迭代对象，包含索引和值的元组，索引为template_id，值为平均完成度
    # Series对象没有itterrows()方法，iterrows()方法是DataFrame对象的方法，返回一个生成器，每次迭代返回一个元组，包含索引和行数据，行数据是一个Series对象
    # print(poor_templates.items())

    # loc方法中的行索引要与DataFrame的索引匹配，列名要与DataFrame的列名匹配，返回一个Series对象，如果dataframe中没有对应的行索引或者列名，则会报错
    # print([template_data.set_index('template_id').loc[template_id, 'template_name'] for template_id in poor_templates.index])
    print([template_data.set_index('template_id').loc[poor_templates.index, 'template_name']])

    if len(poor_templates) > 0:
        insights.append(f"优化建议: {len(poor_templates)} 个模板完成度较低，需要优化")
        for template_id, completion_rate in poor_templates.items():
            template_name = template_data.set_index('template_id').loc[template_id, 'template_name'] # loc通过行索引+列名定位值
            insights.append(f"  ·模板 {template_name} ({template_id}) 平均完成度 {completion_rate:.1f}%")
    
    for insight in insights:
        print(f"• {insight}")

# 数据可视化
def create_visualizations(user_data, resume_data):
        """创建可视化图表"""
        print("\n=== 生成数据可视化图表 ===")
        
        # 创建图表
        fig, axes = plt.subplots(2, 2, figsize=(15, 12))
        fig.suptitle('老鱼简历数据分析报告', fontsize=16, fontweight='bold')
        
        # 1. 用户注册趋势——折线图
        monthly_data = user_data.groupby(
            user_data['registration_date'].dt.to_period('M')
        ).size()
        axes[0, 0].plot(range(len(monthly_data)), monthly_data.values, marker='o')
        axes[0, 0].set_title('月度用户注册趋势')
        axes[0, 0].set_xlabel('月份')
        axes[0, 0].set_ylabel('注册用户数')
        axes[0, 0].grid(True, alpha=0.3)
        
        # 2. 用户类型分布——柱状图
        user_type_counts = user_data['user_type'].value_counts()
        axes[0, 1].pie(user_type_counts.values, labels=user_type_counts.index, autopct='%1.1f%%')
        axes[0, 1].set_title('用户类型分布')
        
        # 3. 简历完成度分布
        axes[1, 0].hist(resume_data['completion_rate'], bins=20, alpha=0.7, color='skyblue')
        axes[1, 0].set_title('简历完成度分布')
        axes[1, 0].set_xlabel('完成度 (%)')
        axes[1, 0].set_ylabel('简历数量')
        axes[1, 0].grid(True, alpha=0.3)
        
        # 4. 模板使用热力图
        template_usage = resume_data.groupby(['template_id', resume_data['created_date'].dt.month]).size().unstack(fill_value=0)
        
        if len(template_usage.columns) > 1:
            sns.heatmap(template_usage.iloc[:10], annot=True, fmt='d', cmap='YlOrRd', ax=axes[1, 1])
            axes[1, 1].set_title('模板月度使用热力图')
            axes[1, 1].set_xlabel('月份')
            axes[1, 1].set_ylabel('模板ID')
        
        plt.tight_layout()
        plt.savefig('resume_analysis_report.png', dpi=300, bbox_inches='tight')
        print("可视化图表已保存为 resume_analysis_report.png")
        
        # 显示图表
        plt.show()

# 运行完整分析
sample_data = generate_sample_data()
sample_data_u = sample_data['user_data']
sample_data_r = sample_data['resume_data']
sample_data_t = sample_data['template_data']

# user_behavior_analysis(sample_data['user_data'])
# resume_creation_analysis(sample_data['resume_data'], sample_data['user_data'], sample_data['template_data'])
# template_performance_analysis(sample_data['resume_data'], sample_data['template_data'])
generate_insights(sample_data['user_data'], sample_data['resume_data'], sample_data['template_data'])
# create_visualizations(sample_data['user_data'], sample_data['resume_data'])
```
